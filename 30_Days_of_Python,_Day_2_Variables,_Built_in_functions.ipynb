```python
#Day 2: 30 Days of Python Programming

first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
country = "England"
city = "London"
age = 30
year = 2026
is_married = False
is_true = True
is_light_on = True
#multiple variables
car, has_license, is_employed = "Toyota", True, True

print(type(first_name))  # <class 'str'>
print(type(last_name))   # <class 'str'>
print(type(full_name))   # <class 'str'>
print(type(country))     # <class 'str'>
print(type(city))        # <class 'str'>
print(type(age))         # <class 'int'>
print(type(year))        # <class 'int'>
print(type(is_married))  # <class 'bool'>
print(type(is_true))     # <class 'bool'>
print(type(is_light_on)) # <class 'bool'>
print(type(car))         # <class 'str'>
print(type(has_license)) # <class 'bool'>
print(type(is_employed)) # <class 'bool'>

print(len(first_name))  # 4
print(len(first_name) > len(last_name))  # True
num_one = 5
num_two = 4
variable_total = num_one + num_two
variable_diff = num_one - num_two
variable_product = num_one * num_two
variable_division = num_one / num_two
variable_remainder = num_one % num_two
variable_exp = num_one ** num_two
floor_division = num_one // num_two
radius = 30
area_of_circle = 3.14 * radius ** 2
circum_of_circle = 2 * 3.14 * radius
print(variable_remainder)  # 1
print(variable_exp)  # 625
print(floor_division)  # 1
print(area_of_circle)  # 2826.0
print(circum_of_circle)  # 188.4
```
